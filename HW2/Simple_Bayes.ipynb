### T2. Plot the posteriors values of the two classes on the same axis. Using the likelihood ratio test, what is the decision boundary for this classifier? Assume equal prior probabilities.

### T3. What happen to the decision boundary if the cat is happy with a prior of 0.75?

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

x = np.linspace(-5, 10, 1000)
l1 = norm.pdf(x, 4, np.sqrt(2))
l2 = norm.pdf(x, 0, np.sqrt(2))

def plot_posteriors(p1, p2, title):
   post1 = (l1 * p1) / (l1 * p1 + l2 * p2)
   post2 = (l2 * p2) / (l1 * p1 + l2 * p2)
    
   plt.figure(figsize=(8, 4))
   plt.plot(x, post1, label='Happy Cat ($w_1$)')
   plt.plot(x, post2, label='Sad Cat ($w_2$)')
    
   boundary = x[np.argmin(np.abs(post1 - post2))]
   plt.axvline(boundary, color='k', linestyle='--', label=f'Boundary: {boundary:.2f}')
    
   plt.title(title)
   plt.legend()
   plt.show()
   return boundary

boundary_t2 = plot_posteriors(0.5, 0.5, "T2")
boundary_t3 = plot_posteriors(0.75, 0.25, "T3")

print(f"Boundary T2: {boundary_t2:.2f}")
print(f"Boundary T3: {boundary_t3:.2f}")
```
